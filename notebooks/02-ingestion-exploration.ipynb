```python
conn.sql("""
SELECT
    b.title AS book_title,
    a.name AS question_author,
    a.text AS question,
    a._dlt_id AS question_id
FROM books_data.books AS b
JOIN books_data.books__archive AS a
    ON b._dlt_id = a._dlt_parent_id
LIMIT 10
""").show()
```

```python
conn.sql("""
SELECT
    a.text AS question,
    r.name AS reply_author,
    r.text AS reply
FROM books_data.books__archive AS a
JOIN books_data.books__archive__replies AS r
    ON a._dlt_id = r._dlt_parent_id
LIMIT 10
""").show()
```

```python
conn.sql("""
SELECT COUNT(*)
FROM books_data.books__archive AS a
LEFT JOIN books_data.books__archive__replies AS r
    ON a._dlt_id = r._dlt_parent_id
WHERE r._dlt_id IS NULL
""").show()
```

```python
conn.sql("""
SELECT
    COUNT(r._dlt_id) AS reply_count,
    COUNT(*) AS number_of_discussions
FROM books_data.books__archive AS a
LEFT JOIN books_data.books__archive__replies AS r
    ON a._dlt_id = r._dlt_parent_id
GROUP BY a._dlt_id
ORDER BY reply_count DESC
LIMIT 10
""").show()
```

```python
conn.sql("""
SELECT
    a.name,
    a.text AS question,
    COUNT(r._dlt_id) AS reply_count
FROM books_data.books__archive AS a
LEFT JOIN books_data.books__archive__replies AS r
    ON a._dlt_id = r._dlt_parent_id
GROUP BY a._dlt_id, a.name, a.text
ORDER BY reply_count DESC
LIMIT 10
""").show()
```

```python
conn.sql("""
SELECT
    MIN(LENGTH(text)) AS smallest_reply,
    MAX(LENGTH(text)) AS largest_reply,
    AVG(LENGTH(text)) AS average_reply
FROM books_data.books__archive__replies
""").show()
```

```python
conn.sql("""
SELECT
    MIN(LENGTH(text)) AS smallest_question,
    MAX(LENGTH(text)) AS largest_question,
    AVG(LENGTH(text)) AS average_question
FROM books_data.books__archive
""").show()
```
