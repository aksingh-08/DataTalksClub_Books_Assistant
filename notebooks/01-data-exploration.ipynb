```python
valid_files = []
failed_files = []

for book_file in book_files:
    try:
        book = frontmatter.load(book_file)

        valid_files.append(book_file)

    except Exception as e:
        failed_files.append({
            "file": book_file.name,
            "error": str(e)
        })
```

```python
print("Total files:", len(book_files))
print("Valid files:", len(valid_files))
print("Failed files:", len(failed_files))
```

```python
all_keys = set()

for book_file in valid_files:
    book = frontmatter.load(book_file)
    all_keys.update(book.metadata.keys())

sorted(all_keys)
```

```python
from collections import Counter

key_counts = Counter()

for book_file in valid_files:
    book = frontmatter.load(book_file)
    key_counts.update(book.metadata.keys())

key_counts
```

```python
missing_titles = []
missing_authors = []
for book_file in book_files:
    book = frontmatter.load(book_file)
    if not book.metadata.get('title'):
        missing_title.append(book_file.name)
    if not book.metadata.get('authors'):
        missing_authors.append(book_file.name)
print('Missing titles:', len(missing_titles))
print('Missing authors:', len(missing_authors))
```

```python
content_lengths= []
for book_file in book_files:
    book = frontmatter.load(book_file)
    content_lengths.append({
    'file': book_file.name,
    'title': book['title'],
    'characters': len(book.content)
    })
```

```python
content_lengths_sorted = sorted(
    content_lengths,
    key=lambda x: x['characters'],
    reverse=True
)
```

```python
content_lengths_sorted[:10]
```

```python
lengths = [
    item['characters']
    for item in content_lengths
]
print('number of books:', len(lengths))
print('smallest:', min(lengths))
print('largest:', max(lengths))
print('average:', sum(lengths) / len(lengths))
```
